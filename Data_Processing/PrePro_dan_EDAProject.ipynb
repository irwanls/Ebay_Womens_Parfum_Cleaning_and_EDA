# Cleanning untuk available, availableText, dan sold

## Memisahkan missing value "availableText" menjadi dataframe baru

```python
df_discontinue = df[df["availableText"].isna()].copy()
df = df[df["availableText"].notna()].copy()
df_discontinue
```

## Mengatasi missing value "available"

### Mengisi missing value jika pada "availableText" terdapat text "Last One"

```python
df.loc[
    df["available"].isna() &
    df["availableText"].str.contains("Last One", case=False, na=False),
    'available'
] = 1
```

### Mengisi missing value jika pada "availableText" terdapat text "Out of Stock"

```python
df.loc[
    df["available"].isna() &
    df["availableText"].str.contains("Out of Stock", case=False, na=False),
    "available"
] = 0
```

### Mengisi missing value jika pada "availableText" terdapat jumlah available-nya

```python
mask = df["available"].isna()

# untuk "n available"
available_extract = (
    df.loc[mask, "availableText"]
    .str.extract(r'(\d+)\s+available', expand=False)
)

df.loc[mask, "available"] = (
    df.loc[mask, 'available']
    .fillna(pd.to_numeric(available_extract, errors='coerce'))
)

# untuk lot dan jumlah per-lot
lots = df.loc[mask, 'availableText'].str.extract(
    r'(\d+)\s+lots?\s+available',
    expand=False
)

items_per_lot = df.loc[mask, 'availableText'].str.extract(
    r'(\d+)\s+items?\s+per\s+lot',
    expand=False
)

total_items = (
    pd.to_numeric(lots, errors='coerce') *
    pd.to_numeric(items_per_lot, errors='coerce')
)

df.loc[mask, 'available'] = (
    df.loc[mask, 'available']
    .fillna(total_items)
)

# untuk bahasa spanyol
mask_spanish = (
    df['available'].isna() &
    df['availableText'].str.contains(
        r'\d+\s+disponibles?',
        case=False,
        na=False
    )
)

df.loc[mask_spanish, 'available'] = (
    df.loc[mask_spanish, 'availableText']
    .str.extract(r'(\d+)\s+disponibles?', expand=False)
    .astype(int)
)

# input manual
df.loc[990, 'available'] = 32
```

### Mengisi missing value dengan median untuk Limited quantity

```python
mask_limited = (
    df['available'].isna() &
    df['availableText'].str.contains(
        'Limited quantity available',
        case=False,
        na=False
    )
)

df.loc[mask_limited, 'available'] = df['available'].median()
```

# Pengecekan outlier

## Outlier kolom type

```python
type_counts = df["type"].value_counts()

Q1 = type_counts.quantile(0.25)
Q3 = type_counts.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

type_outlier = type_counts[
    (type_counts < lower) |
    (type_counts > upper)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
type_outlier
```

### Mengelompokan outlier ke kategori baru

```python
df["type"].value_counts()
```

```python
type_counts = df["type"].value_counts()
df["type"] = df["type"].apply(
    lambda x: x if type_counts[x] > 10 else "Others")
        
df["type"].value_counts()
```

## Outlier kolom price

```python
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

price_outlier = df[
    (df["price"] < lower) |
    (df["price"] > upper)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
price_outlier["price"].count()
```
